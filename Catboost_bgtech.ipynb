#### Cramer's V

Esta é uma medida estatística que avalia a força de associação entre duas variáveis categóricas, variando de 0 (nenhuma associação) a 1 (associação perfeita). Ele é derivado do teste qui-quadrado e é usado em tabelas de contingência para determinar se a relação entre as categorias é real ou devido ao acaso.

```python
def cramers_v(x, y):
    """
    utiliza Cramer's V statistic para calcular a correlação entre variáveis categóricas

    Parameters:
    x, y: pandas.Series
        variáveis categóricas para comparação

    Returns: float
        valor entre 0 e 1, onde 1 = correlação perfeita e 0 = sem correlação
    """

    confusion_matrix = pd.crosstab(x, y)     # Cria a tabela de contingência entre as duas variáveis


    chi2 = chi2_contingency(confusion_matrix)[0]    # Calcula o valor do qui-quadrado (chi2) a partir da tabela


    n = confusion_matrix.sum().sum()    # Número total de observações


    phi2 = chi2 / n    # Phi² não corrigido


    r, k = confusion_matrix.shape    # Número de linhas (r) e colunas (k) da tabela


    phi2corr = max(0, phi2 - (k-1)*(r-1)/(n-1))    # Correção de phi² para reduzir viés em tabelas pequenas


    rcorr = r - (r-1)**2/(n-1)    # Correção para o número efetivo de linhas


    kcorr = k - (k-1)**2/(n-1)    # Correção para o número efetivo de colunas


    return np.sqrt(phi2corr / min((kcorr-1), (rcorr-1)))    # Retorna o Cramér's V (entre 0 e 1)
```

#### Remoção das features altamente correlacionadas


```python
import numpy as np
import pandas as pd
from sklearn.base import BaseEstimator, TransformerMixin
import scipy.stats as ss

def cramers_v(x, y):
    confusion_matrix = pd.crosstab(x, y)
    chi2 = ss.chi2_contingency(confusion_matrix)[0]
    n = confusion_matrix.sum().sum()
    phi2 = chi2 / n
    r, k = confusion_matrix.shape
    phi2corr = max(0, phi2 - ((k-1)*(r-1))/(n-1))
    rcorr = r - ((r-1)**2)/(n-1)
    kcorr = k - ((k-1)**2)/(n-1)
    return np.sqrt(phi2corr / min((kcorr-1), (rcorr-1)))

class DropCorr(BaseEstimator, TransformerMixin):
    def __init__(self, corr_threshold=0.8, top_n=10):
        self.corr_threshold = corr_threshold
        self.to_drop_ = []
        self.high_corr_pairs_ = []
        self.top_n = top_n

    def fit(self, X, y=None):
        to_drop = set()
        cols = X.columns.tolist()
        corr_pairs = []

        for i, col1 in enumerate(cols):
            if col1 in to_drop:
                continue
            for col2 in cols[i+1:]:
                if col2 in to_drop:
                    continue

                # Caso 1: ambos numéricos
                if np.issubdtype(X[col1].dtype, np.number) and np.issubdtype(X[col2].dtype, np.number):
                    corr = X[col1].corr(X[col2])
                    if abs(corr) > self.corr_threshold:
                        to_drop.add(col2)
                        corr_pairs.append((col1, col2, corr))

                # Caso 2: ambos categóricos
                elif (X[col1].dtype == "object" or isinstance(X[col1].dtype, pd.CategoricalDtype)):
                    if (X[col2].dtype == "object" or isinstance(X[col2].dtype, pd.CategoricalDtype)):
                        v = cramers_v(X[col1], X[col2])
                        if v > self.corr_threshold:
                            to_drop.add(col2)
                            corr_pairs.append((col1, col2, v))

        # salva lista
        self.to_drop_ = list(to_drop)

        # salva top N correlações
        self.high_corr_pairs_ = sorted(corr_pairs, key=lambda x: abs(x[2]), reverse=True)

        # printa no fit
        print("\nTop correlações encontradas:")
        for col1, col2, val in self.high_corr_pairs_[:self.top_n]:
            print(f"{col1} ↔ {col2}: {val:.3f}")

        return self

    def transform(self, X):
        return X.drop(columns=self.to_drop_, errors="ignore")

    def get_features_removed(self):
        return self.to_drop_

    def get_high_corr_pairs(self):
        return self.high_corr_pairs_
```

#### Dropped dataframe

```python
# Definir o threshold de correlação
corr_threshold = 0.75

# Instanciar o transformador
drop_corr = DropCorr(corr_threshold=corr_threshold)

# "Treina" o transformador no dataframe
drop_corr.fit(df)

# Transforma o dataframe removendo as colunas correlacionadas
df_dropped = drop_corr.transform(df)

# Recupera a lista de colunas removidas
dropped = drop_corr.get_features_removed()
```

### TRATAMENTO DE OUTLIERS

```python
def winsorize_iqr(df, cols=None, k=1.5, exclude=None):
    """
    Aplica winsorização com base no IQR para as colunas numéricas escolhidas,
    ignorando colunas definidas no parâmetro exclude.

    Parameters:
    df : pandas.DataFrame
        Dataframe de entrada
    cols : list, optional
        Lista de colunas a serem tratadas (se None, aplica em todas numéricas)
    k : float, optional (default=1.5)
        Fator multiplicador do IQR para definir os limites
    exclude : list, optional
        Lista de colunas a serem ignoradas (default = ["pdv", "zipcode", "produto"])

    Returns:
    df_winsor : pandas.DataFrame
        Dataframe com winsorização aplicada
    """

    df_winsor = df.copy()

    if exclude is None:
        exclude = ["pdv", "zipcode", "produto"]

    if cols is None:
        cols = df_winsor.select_dtypes(include=np.number).columns.tolist()

    # remove colunas a excluir
    cols = [c for c in cols if c not in exclude]

    for col in cols:
        Q1 = df_winsor[col].quantile(0.25)
        Q3 = df_winsor[col].quantile(0.75)
        IQR = Q3 - Q1

        lower = Q1 - k * IQR
        upper = Q3 + k * IQR

        # Winsoriza
        df_winsor[col] = np.where(df_winsor[col] < lower, lower,
                           np.where(df_winsor[col] > upper, upper, df_winsor[col]))

    return df_winsor
```

### REPRESENTATIVIDADE DAS VARIÁVEIS CATEGÓRICAS

O intuito desta etapa é identificar as categorias “pequenas” ou de baixa representatividade, agregando-as em grupos e calculando as médias de quantidade por categoria, o que irá facilitar a execução do modelo, uma vez que será reduzido o número de features

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display

colunas = [
    'premise',
    'categoria_pdv',
    'categoria',
    'label',
    'marca']

for i in colunas:
    print(f'Análise para a coluna: "{i}"')

    # Passo 1: Calcular a representatividade das categorias
    representatividade = merged_df[i].value_counts(normalize=True, dropna =False).reset_index()
    representatividade.columns = [i, 'representatividade_%']
    representatividade['representatividade_%'] = representatividade['representatividade_%'] * 100

    # Passo 2: Calcular a média da quantidade para cada categoria
    media_quantidades = merged_df.groupby(i)['quantity'].mean().reset_index()
    media_quantidades.columns = [i, 'media_quantidade']

    # Passo 3: Juntar os dois DataFrames
    df_analise = pd.merge(representatividade, media_quantidades, on=i, how='left')

    # Passo 4: Exibir o DataFrame consolidado
    print('\nRepresentatividade e Média de Quantidade por Categoria:')
    display(df_analise)

    print('-' * 50) # Adiciona uma linha divisória para melhor visualização

    # O código para o gráfico pode ser mantido ou ajustado
    # para usar o novo DataFrame 'df_analise'

    # Criando o gráfico com seaborn (usando o df_analise para consistência)
    plt.figure(figsize=(10, 6))
    sns.barplot(x=df_analise[i], y=df_analise['representatividade_%'])
    plt.title(f'Representatividade das categorias de "{i}"', fontsize=16)
    plt.xlabel('Categorias', fontsize=12)
    plt.ylabel('Porcentagem (%)', fontsize=12)
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()

    print('-' * 50)
```

### CatBoost

O CatBoost é uma biblioteca avançada de reforço de gradiente, projetada especificamente para lidar com os desafios de lidar com dados categóricos em machine learning. Nessa ferramenta, a transformação de recursos para dados categóricos e de texto é automatizada, e constrói árvores de decisão usando otimização baseada em gradiente, ou seja, as árvores são ajustadas ao gradiente negativo da função de perda 

```python
#Instalação do catboost
!pip install catboost --quiet
```

```python
import pandas as pd
import numpy as np
from catboost import CatBoostRegressor, Pool
import os
import math

# ===============================
# 1. Preparação dos dados
# ===============================

TARGET = "quantity"

X = df_dropped.drop(columns=[TARGET])
y = df_dropped[TARGET]

categorical_features = X.select_dtypes(include=["object", "category"]).columns.tolist()

for col in X.columns:
    if X[col].dtype != "object":
        if X[col].apply(lambda x: isinstance(x, str)).any():
            X[col] = X[col].astype(str)
            categorical_features.append(col)

for col in categorical_features:
    X[col] = X[col].fillna("missing").astype(str)

# ===============================
# 2. Split treino/validação manual
# ===============================

np.random.seed(42)
msk = np.random.rand(len(X)) < 0.8
X_train, y_train = X[msk].copy(), y[msk].copy()
X_val, y_val = X[~msk].copy(), y[~msk].copy()

train_pool = Pool(X_train, y_train, cat_features=categorical_features)
val_pool = Pool(X_val, y_val, cat_features=categorical_features)

# ===============================
# 3. Ensemble CatBoost
# ===============================

model1 = CatBoostRegressor(
    iterations=1000,
    learning_rate=0.05,
    depth=6,
    eval_metric='RMSE',
    random_seed=42,
    early_stopping_rounds=50,
    verbose=100
)
model1.fit(train_pool, eval_set=val_pool)

model2 = CatBoostRegressor(
    iterations=1000,
    learning_rate=0.03,
    depth=8,
    eval_metric='RMSE',
    random_seed=99,
    early_stopping_rounds=50,
    verbose=100
)
model2.fit(train_pool, eval_set=val_pool)

preds_val1 = model1.predict(X_val)
preds_val2 = model2.predict(X_val)
y_pred_val = (preds_val1 + preds_val2) / 2
rmse = np.sqrt(((y_val - y_pred_val) ** 2).mean())
print(f"📊 RMSE de validação (ensemble CatBoost): {rmse:.2f}")

# ===============================
# 4. Previsão em chunks para evitar MemoryError
# ===============================

weeks = list(range(1,6))
pairs = df_dropped[["pdv","produto"]].drop_duplicates().reset_index(drop=True)
n_pairs = len(pairs)
chunk_pairs = 5000  # mais seguro

num_features = X.drop(columns=categorical_features).columns.tolist()
df_features = df_dropped.groupby(["pdv","produto"], as_index=False)[num_features].mean(numeric_only=True)

output_file = "previsoes_parcial.csv"
final_file = "previsoes.csv"

# remove arquivos antigos
for f in [output_file, final_file]:
    if os.path.exists(f):
        os.remove(f)

total_chunks = math.ceil(n_pairs / chunk_pairs)
train_cols = list(X_train.columns)
preds_final = []

for i in range(total_chunks):
    start = i * chunk_pairs
    end = min(start + chunk_pairs, n_pairs)
    chunk_pairs_df = pairs.iloc[start:end].copy()

    # expandir para semanas
    reps = len(weeks)
    pdv_arr = np.repeat(chunk_pairs_df["pdv"].values, reps)
    prod_arr = np.repeat(chunk_pairs_df["produto"].values, reps)
    week_arr = np.tile(weeks, len(chunk_pairs_df))

    df_chunk = pd.DataFrame({
        "semana": week_arr,
        "pdv": pdv_arr,
        "produto": prod_arr
    })

    # merge com features numéricas médias
    df_chunk = df_chunk.merge(df_features, on=["pdv","produto"], how="left")
    for col in num_features:
        if col in df_chunk.columns:
            med = X[col].median() if col in X.columns else 0
            df_chunk[col] = df_chunk[col].fillna(med)

    for col in categorical_features:
        if col in df_chunk.columns:
            df_chunk[col] = df_chunk[col].fillna("missing").astype(str)
        else:
            df_chunk[col] = "missing"

    for c in train_cols:
        if c not in df_chunk.columns:
            df_chunk[c] = 0

    X_chunk = df_chunk[train_cols].copy()

    # predição ensemble
    preds1 = model1.predict(X_chunk)
    preds2 = model2.predict(X_chunk)
    preds = (preds1 + preds2) / 2.0
    df_chunk["quantidade"] = np.round(preds).astype(int)
    preds_final.append(df_chunk)

    # salvar parcial
    pd.concat(preds_final).to_csv(output_file, sep=";", index=False, encoding="utf-8")
    print(f"✅ Chunk {i+1}/{total_chunks} salvo (pairs {start}:{end})")

# salvar final
df_final = pd.concat(preds_final, ignore_index=True)[["semana","pdv","produto","quantidade"]]
os.replace(output_file, final_file)
print("🏁 Arquivo final 'previsoes.csv' gerado com sucesso!")

```
